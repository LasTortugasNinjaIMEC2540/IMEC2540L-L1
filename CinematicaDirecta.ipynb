```python
import numpy as np
#Determinación longitudes por pixeles:

#Dimensiones reales [mm]:
#l_real = k*l_pix

O_pix=np.array([-17058, 24338.7])
A_pix=np.array([-17075.5, 24012.4])
B_pix=np.array([-16747.5, 24012.4])
C_pix=np.array([-16747.5, 24140])
D_pix=np.array([-16697.3, 24136.6])

OA_pix = np.linalg.norm(O_pix-A_pix)
k = 130/OA_pix
L1=float(130)

#AB
AB_pix = np.linalg.norm(A_pix-B_pix)
L2=float(k*AB_pix)

#BC
BC_pix = np.linalg.norm(B_pix-C_pix)
L3 = float(k*BC_pix)

#CD
CD_pix = np.linalg.norm(C_pix-D_pix)
L4 = float(k*CD_pix)

display(L1,L2,L3,L4)
```

```python
#FK: [theta1, theta2] -> [x_D, y_D]

import numpy as np
import sympy
from sympy import symbols
from sympy.physics.mechanics import ReferenceFrame, dynamicsymbols, init_vprinting

init_vprinting()

theta1, theta2 = dynamicsymbols("theta1, theta2")
l1, l2, l3, l4, x_D, y_D = symbols("l1, l2, l3, l4, x_D, y_D")

O=ReferenceFrame("O")
A=O.orientnew("A", "Axis", (-theta1, O.z))
B=A.orientnew("B", "Axis", (-theta2, O.z))
rOA = l1*A.y
rAB = l2*B.x
rBC = -l3*O.y #Independiente de angulos
rCD = -l4*O.x #Independiente de angulos

rOD = rOA+rAB+rBC+rCD
rOD.express(O)
x_D_expr = rOD.dot(O.x)
y_D_expr = rOD.dot(O.y)

longitudes = {l1: L1, l2: L2, l3: L3, l4: L4}
x_D_expr_s = rOD.dot(O.x).subs(longitudes)
y_D_expr_s = rOD.dot(O.y).subs(longitudes)

x_D_fun = sympy.lambdify((theta1, theta2), x_D_expr_s)
y_D_fun = sympy.lambdify((theta1, theta2), y_D_expr_s)

def FK(theta1, theta2):
    return (x_D_fun(theta1, theta2), y_D_fun(theta1, theta2))

display(sympy.Eq(x_D, x_D_expr))
display(sympy.Eq(y_D, y_D_expr))
```
